```python
import os

def get_op_path(op_num):
    """
    Returns the folder path for a given OP number.

    Example:
    3 -> ./storeOps/OP03
    """
    
    base_path = os.path.join(".", "storeOps")
    
    op_str = f"OP{int(op_num):02d}"
    path = os.path.join(base_path, op_str)
    
    if not os.path.exists(path):
        raise FileNotFoundError(f"{path} does not exist")
    
    return path
```

```python
print(get_op_path(3))
```

```python
import pandas as pd

def load_input_signals(op_num):
    """
    Loads the input signals of a given OP.
    Returns a dictionary with all values.
    """
    
    folder = get_op_path(op_num)
    
    # ✅ FIX: ohne führende 0 im Dateinamen
    file_name = f"OP{int(op_num)}_Input Signale.csv"
    
    file_path = os.path.join(folder, file_name)
    
    if not os.path.exists(file_path):
        raise FileNotFoundError(f"{file_path} does not exist")
    
    df = pd.read_csv(file_path)
    
    if df.shape[0] == 0:
        raise ValueError("File is empty")
    
    data = df.iloc[0].to_dict()
    
    return data
```

```python
inputs = load_input_signals(3)

print(inputs)
```

```python
def get_input_value(op_num, header_name):
    """
    Returns the value of a specific input signal (by column name).

    Parameters
    ----------
    op_num : int
    header_name : str

    Returns
    -------
    value
    """
    
    data = load_input_signals(op_num)
    
    if header_name not in data:
        raise KeyError(f"{header_name} not found in input signals")
    
    return data[header_name]
```

```python
value = get_input_value(3, "Cell Current Monitor (A)")
print(value)
```

```python
import os
import pandas as pd

def load_fluid_properties(op_num):
    """
    Loads the fluid properties of a given OP.
    
    Returns:
    dict with all fluid values
    """
    
    folder = get_op_path(op_num)
    
    # ✅ gleiches Namensproblem beachten (ohne führende 0!)
    file_name = f"OP{int(op_num)}_Fluidstoffwerte.csv"
    
    file_path = os.path.join(folder, file_name)
    
    if not os.path.exists(file_path):
        raise FileNotFoundError(f"{file_path} does not exist")
    
    df = pd.read_csv(file_path)
    
    if df.shape[0] == 0:
        raise ValueError("File is empty")
    
    data = df.iloc[0].to_dict()
    
    return data
```

```python
def get_fluid_value(op_num, header_name):
    """
    Returns a specific fluid property by column name.
    """
    
    data = load_fluid_properties(op_num)
    
    if header_name not in data:
        raise KeyError(f"{header_name} not found in fluid properties")
    
    return data[header_name]
```

```python
value = get_fluid_value(3, "Density Monitor (kg/m^3)")
print(value)
```
