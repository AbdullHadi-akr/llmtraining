```python
import os

def get_op_path(op_num):
    """
    Returns the folder path for a given OP number.

    Example:
    3 -> ./storeOps/OP03
    """
    
    base_path = os.path.join(".", "storeOps")
    
    op_str = f"OP{int(op_num):02d}"
    path = os.path.join(base_path, op_str)
    
    if not os.path.exists(path):
        raise FileNotFoundError(f"{path} does not exist")
    
    return path
```

```python
import pandas as pd
import os

def load_FMU1(op_num):
    """
    Loads the full FMU1 dataframe for a given OP.
    Returns the whole DataFrame.
    """
    
    folder = get_op_path(op_num)
    file_name = f"OP{int(op_num)}_Batemo FMU1.csv"
    file_path = os.path.join(folder, file_name)
    
    if not os.path.exists(file_path):
        raise FileNotFoundError(f"{file_path} does not exist")
    
    # robust CSV reading
    df = pd.read_csv(file_path, sep=",", engine="python")
    
    if df.shape[0] == 0:
        raise ValueError("File is empty")
    
    return df
```

```python
import pandas as pd
import os

def get_FMU1_value(op_num, header_name, row):
    """
    Returns the value for a given header (column) and row.
    
    Parameters:
        op_num (int): OP number
        header_name (str): column name (e.g. "ac_V_an Monitor")
        row (int): row index (default = 0)
    
    Returns:
        value from dataframe
    """

    folder = get_op_path(op_num)
    file_name = f"OP{int(op_num)}_Batemo FMU1.csv"
    file_path = os.path.join(folder, file_name)

    if not os.path.exists(file_path):
        raise FileNotFoundError(f"{file_path} does not exist")

    df = pd.read_csv(file_path, sep=",", engine="python")

    if df.empty:
        raise ValueError("File is empty")

    if header_name not in df.columns:
        raise ValueError(f"Column '{header_name}' not found")

    return df[header_name].iloc[row]
```

```python
import pandas as pd
import os

def load_FMU2(op_num):
    """
    Loads the full FMU2 dataframe for a given OP.
    """

    folder = get_op_path(op_num)
    file_name = f"OP{int(op_num)}_Batemo FMU2.csv"
    file_path = os.path.join(folder, file_name)

    if not os.path.exists(file_path):
        raise FileNotFoundError(f"{file_path} does not exist")

    df = pd.read_csv(file_path, sep=",", engine="python")

    if df.empty:
        raise ValueError("File is empty")

    return df
```

```python
def get_FMU2_value(op_num, header_name, row):
    """
    Returns a value from FMU2 by column (header) and row.

    Example headers:
    "ac_V_an2 Monitor"
    "bc_I_2 Monitor"
    "bc_OCV_2 Monitor"
    ...
    """

    df = load_FMU2(op_num)

    if header_name not in df.columns:
        raise ValueError(f"Column '{header_name}' not found")

    return df[header_name].iloc[row]
```
